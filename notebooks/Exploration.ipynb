```python
import pandas as pd

# 1. Chargement des données (XGBoost a déjà fait le travail)
df = pd.read_csv("../data/meta_dataset.csv")

print("📊 --- ANALYSE DE LA CONFIANCE INTERNE (XGBOOST SEUL) ---")
print("Rappel : Objectif de rentabilité avec frais > 52.0% (Ratio 1:1)\n")

# On teste la confiance de XGBoost (de 50% à 70% de certitude)
seuils_xgboost = [0.50, 0.52, 0.55, 0.58, 0.60, 0.62, 0.65]

for seuil in seuils_xgboost:
    # On filtre les trades où XGBoost était sûr à plus de X%
    trades_filtres = df[df['primary_proba'] > seuil]
    
    total_trades = len(trades_filtres)
    
    if total_trades > 0:
        # meta_target = 1 (Succès), 0 (Échec)
        gagnants = trades_filtres['meta_target'].sum()
        win_rate = (gagnants / total_trades) * 100
        
        if win_rate > 52.0:
            etat = "✅ RENTABLE"
        else:
            etat = "❌ PERTE"
            
        print(f"Confiance > {seuil*100:.0f}% -> Win Rate: {win_rate:.2f}% | Trades pris: {total_trades} | {etat}")
    else:
        print(f"Confiance > {seuil*100:.0f}% -> 0 trade pris.")
```

```python
import pandas as pd

# 1. Chargement du nouveau dataset (filtré par la SMA 50)
df = pd.read_csv("../data/meta_dataset.csv")

print("📊 --- ANALYSE DE LA CONFIANCE INTERNE (XGBOOST) ---")
print("Objectif réel : > 45.0% pour être RENTABLE AVEC frais (Ratio 1.5:1)\n")

seuils_xgboost = [0.50, 0.52, 0.54, 0.56, 0.58, 0.60, 0.62]

for seuil in seuils_xgboost:
    trades_filtres = df[df['primary_proba'] > seuil]
    total_trades = len(trades_filtres)
    
    if total_trades > 0:
        gagnants = trades_filtres['meta_target'].sum()
        win_rate = (gagnants / total_trades) * 100
        
        if win_rate > 45.0:
            etat = "✅ RENTABLE (Survit aux frais)"
        elif win_rate > 40.0:
            etat = "⚠️ BREAK-EVEN"
        else:
            etat = "❌ PERTE"
            
        print(f"Confiance > {seuil*100:.0f}% -> Win Rate: {win_rate:.2f}% | Trades: {total_trades} | {etat}")
    else:
        print(f"Confiance > {seuil*100:.0f}% -> 0 trade pris.")
```

```python
import pandas as pd
import numpy as np

# 1. Chargement du dataset point-à-point
df = pd.read_csv("../data/meta_dataset.csv")

print("📊 --- ANALYSE DE LA DIRECTION PURE (XGBOOST À 16h00) ---")
print("Objectif de base : Avoir raison > 50.0% du temps (Net de frais)\n")

seuils_xgboost = [0.50, 0.52, 0.54, 0.56, 0.58, 0.60]

for seuil in seuils_xgboost:
    trades_filtres = df[df['primary_proba'] > seuil]
    total_trades = len(trades_filtres)
    
    if total_trades > 0:
        gagnants = trades_filtres['meta_target'].sum()
        win_rate = (gagnants / total_trades) * 100
        
        if win_rate > 50.0:
            etat = "✅ AVANTAGE STATISTIQUE"
        else:
            etat = "❌ HASARD / BRUIT"
            
        print(f"Confiance > {seuil*100:.0f}% -> Win Rate: {win_rate:.2f}% | Trades: {total_trades} | {etat}")
    else:
        print(f"Confiance > {seuil*100:.0f}% -> 0 trade pris.")
```
